```python
# 1. สร้าง class Student
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

# 2. เริ่มต้นด้วยรายชื่อ 3-5 คน
students = [
    Student("Alice", 85),
    Student("Bob", 60),
    Student("Charlie", 45),
    Student("David", 72),
    Student("Eve", 90)
]

# 3. ให้ผู้ใช้เพิ่มข้อมูลนักเรียนใหม่ได้ (หยุดเมื่อพิมพ์ stop)
while True:
    name_input = input("ชื่อนักเรียน (พิมพ์ 'stop' เพื่อหยุด): ")
    if name_input.lower() == "stop":
        break
    try:
        score_input = int(input("คะแนนสอบ: "))
        students.append(Student(name_input, score_input))
    except ValueError:
        print("กรุณากรอกคะแนนเป็นตัวเลข")

# 4. ใช้ lambda + filter เพื่อแสดงรายชื่อที่ได้คะแนนน้อยกว่า 50
print("\nนักเรียนที่ได้คะแนนน้อยกว่า 50:")
low_scores = list(filter(lambda s: s.score < 50, students))
for s in low_scores:
    print(f"{s.name}: {s.score}")

# 5. ใช้ lambda + map เพื่อแปลงคะแนนเป็นเกรด
print("\nรายชื่อนักเรียนพร้อมเกรด:")
def get_grade(score):
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

graded_students = list(map(lambda s: (s.name, get_grade(s.score)), students))
for name, grade in graded_students:
    print(f"{name}: Grade {grade}")

# 6. ใช้ lambda + sorted เพื่อเรียงคะแนนจากมากไปน้อย
print("\nนักเรียนเรียงตามคะแนนจากมากไปน้อย:")
sorted_students = sorted(students, key=lambda s: s.score, reverse=True)
for s in sorted_students:
    print(f"{s.name}: {s.score}")
```
