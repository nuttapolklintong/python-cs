```python
class Car:
    def __init__(self, brand, year):
        self.brand = brand
        self.year = year
    
    def get_info(self):
        return f"{self.brand} - {self.year}"
    
    def update_year(self, new_year):
        self.year = new_year

class Garage:
    def __init__(self):
        self.cars = []  # ใช้สำหรับเก็บรถทั้งหมดในโรงรถ

    def add_car(self, car):
        self.cars.append(car)

    def remove_car(self, brand):
        for car in self.cars:
            if car.brand == brand:
                self.cars.remove(car)
                break

    def find_car(self, brand):
        for i, car in enumerate(self.cars):
            if car.brand == brand:
                return i
        return -1  # ถ้าไม่พบรถ

    def sort_cars(self):
        self.cars.sort(key=lambda car: car.brand)

    def reverse_cars(self):
        self.cars.reverse()

    def clear_all(self):
        self.cars.clear()

    def show_all_cars(self):
        for car in self.cars:
            print(car.get_info())

car1 = Car("Honda", 2020)
car2 = Car("Toyota", 2019)
car3 = Car("Nissan", 2021)

garage = Garage()
garage.add_car(car1)
garage.add_car(car2)
garage.add_car(car3)

print("All cars in the garage:")
garage.show_all_cars()

car_to_update = input("Enter the brand of the car you want to update the year for: ").strip()

# หา car ตามยี่ห้อในโรงรถ
found_car = None
for car in garage.cars:
    if car.brand.lower() == car_to_update.lower():
        found_car = car
        break

# ถ้าพบรถคันที่ต้องการอัปเดต
if found_car:
    try:
        new_year = int(input(f"Enter the new year for {car_to_update}: ").strip())
        found_car.update_year(new_year)
        print(f"Updated {car_to_update} year to {new_year}")
    except ValueError:
        print("Please enter a valid year!")
else:
    print(f"Car with brand {car_to_update} not found in the garage!")

# แสดงผลหลังจากอัปเดต
print("\nAll cars after updating the year:")
garage.show_all_cars()


for car in garage.cars:
    if car.brand.lower() == car_to_update.lower():
        car.update_year(new_year)
        print(f"After update: {car.brand} {car.year}")
        break
else:
    print(f"Car with brand {car_to_update} not found!")


for i in range(2):
    brand = input(f"Enter the brand for car {i + 1}: ").strip()
    year = int(input(f"Enter the year for {brand}: ").strip())
    new_car = Car(brand, year)
    garage.add_car(new_car)

print("\nAll cars in the garage after adding 2 more cars:")
garage.show_all_cars()

print("\nAll cars in the garage after adding more cars:")
garage.show_all_cars()

garage.remove_car("Honda")
print("\nAll cars after removing Honda:")
garage.show_all_cars()

sort_option = input("Sort cars by (brand/year)? ").strip().lower()

if sort_option == "brand":
    garage.sort_cars()
    print("\nAll cars after sorting by brand:")
elif sort_option == "year":
    garage.cars.sort(key=lambda car: car.year)
    print("\nAll cars after sorting by year:")
else:
    print("Invalid option! Cars will not be sorted.")
    print("\nAll cars in the garage:")

reverse_option = input("Do you want to reverse the list of cars? (yes/no): ").strip().lower()

if reverse_option == "yes":
    garage.reverse_cars()
    print("\nAll cars after reversing:")
else:
    print("Cars order remains the same.")

garage.show_all_cars()



garage.clear_all()
print("\nAll cars after clearing the garage:")
garage.show_all_cars()
```
