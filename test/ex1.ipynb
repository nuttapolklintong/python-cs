```python
class Car:
    def __init__(self, brand, year):
        self.brand = brand
        self.year = year

    def get_info(self):
        return f"{self.brand} - {self.year}"

    def update_year(self, new_year):
        self.year = new_year

car1 = Car("Honda", 2020)
car2 = Car("Toyota", 2018)
car3 = Car("BMW", 2022)

print(car1.get_info())  # Honda - 2020
print(car2.get_info())  # Toyota - 2018
print(car3.get_info())  # BMW - 2022

car1.update_year(2023)
print(car1.get_info())  # Honda - 2023
```
