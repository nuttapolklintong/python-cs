```python
class Car:
    def __init__(self, brand, year):
        self.brand = brand
        self.year = year
    
    def get_info(self):
        return f"{self.brand} - {self.year}"
    
    def update_year(self, new_year):
        self.year = new_year

class Garage:
    def __init__(self):
        self.cars = []  # ใช้สำหรับเก็บรถทั้งหมดในโรงรถ

    # 1. add_car(car): เพิ่มรถใหม่เข้าไปใน list
    def add_car(self, car):
        self.cars.append(car)

    # 2. remove_car(brand): ลบรถตามยี่ห้อที่กำหนด
    def remove_car(self, brand):
        for car in self.cars:
            if car.brand == brand:
                self.cars.remove(car)
                break

    # 3. find_car(brand): คืนค่าตำแหน่ง Index ของรถที่ค้นหา
    def find_car(self, brand):
        for i, car in enumerate(self.cars):
            if car.brand == brand:
                return i
        return -1  # ถ้าไม่พบรถ

    # 4. sort_cars(): เรียงลำดับรถตามชื่อยี่ห้อ
    def sort_cars(self):
        self.cars.sort(key=lambda car: car.brand)

    # 5. reverse_cars(): เรียงลำดับกลับ
    def reverse_cars(self):
        self.cars.reverse()

    # 6. clear_all(): ลบข้อมูลรถทั้งหมด
    def clear_all(self):
        self.cars.clear()

    # แสดงข้อมูลทั้งหมดของรถใน garage พร้อม index
    def show_all_cars(self):
        return ', '.join([f"{i}: {car.get_info()}" for i, car in enumerate(self.cars)])  # เพิ่ม index

# ฟังก์ชันหลักเพื่อรับข้อมูลจากผู้ใช้
def main():
    # สร้างออบเจกต์ Garage
    garage = Garage()

    while True:
        print("\n--- Garage Menu ---")
        print("1. Add car")
        print("2. Update car year")
        print("3. Sort cars")
        print("4. Reverse cars")
        print("5. Remove car")
        print("6. Clear all cars")
        print("7. Show all cars")
        print("8. Exit")
        
        choice = input("Enter your choice: ").strip()

        if choice == '1':
            # 1. เพิ่มรถ
            brand = input("Enter car brand: ").strip()
            year = int(input("Enter car year: ").strip())
            new_car = Car(brand, year)
            garage.add_car(new_car)
            print(f"{brand} - {year} added to the garage.")
        
        elif choice == '2':
            # 2. อัปเดตปีของรถ
            car_to_update = input("Enter the brand of the car you want to update the year for: ").strip()
            new_year = int(input(f"Enter the new year for {car_to_update}: ").strip())
            found = False
            for car in garage.cars:
                if car.brand.lower() == car_to_update.lower():
                    car.update_year(new_year)
                    print(f"After update: {car.brand} {car.year}")
                    found = True
                    break
            if not found:
                print(f"Car with brand {car_to_update} not found!")
        
        elif choice == '3':
            # 3. เรียงลำดับรถตามยี่ห้อ
            garage.sort_cars()
            print("Sorted cars:", garage.show_all_cars())
        
        elif choice == '4':
            # 4. เรียงลำดับกลับ
            garage.reverse_cars()
            print("Reversed cars:", garage.show_all_cars())
        
        elif choice == '5':
            # 5. ลบรถตามยี่ห้อ
            brand_to_remove = input("Enter the brand of the car you want to remove: ").strip()
            garage.remove_car(brand_to_remove)
            print(f"{brand_to_remove} removed from the garage.")
        
        elif choice == '6':
            # 6. ลบข้อมูลรถทั้งหมด
            garage.clear_all()
            print("All cars have been cleared from the garage.")
        
        elif choice == '7':
            # 7. แสดงข้อมูลรถทั้งหมด
            print("All cars:", garage.show_all_cars())
        
        elif choice == '8':
            # 8. ออกจากโปรแกรม
            print("Exiting the garage program.")
            break
        
        else:
            print("Invalid choice. Please try again.")

# เรียกใช้ฟังก์ชันหลัก
if __name__ == "__main__":
    main()
```
